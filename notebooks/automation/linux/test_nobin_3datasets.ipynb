# Show4DSTEM - no bin, three datasets (Linux CUDA)

Three full-resolution (192x192) acquisitions eager-stacked into one 5D array (~54 GB on one 96 GB card). The Dataset slider switches between them at full detector resolution. For more no-bin datasets than fit one card, span them with `load(masters, devices=[0, 1])` for capacity.

```python
from quantem.widget import load, Show4DSTEM

masters = [
    "/data/sample/series/Sample_3.6Mx_21.4mrad_185mmcl_20pA_01_master.h5",
    "/data/sample/series/Sample_3.6Mx_21.4mrad_185mmcl_20pA_03_master.h5",
    "/data/sample/series/Sample_3.6Mx_21.4mrad_185mmcl_20pA_11_master.h5",
]
Show4DSTEM(load(masters, det_bin=1))
```
