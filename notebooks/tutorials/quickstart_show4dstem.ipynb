# quantem.widget — Show4DSTEM quickstart

Open and explore a 4D-STEM dataset in your browser, on **any machine** - the backend
auto-detects (NVIDIA CUDA, Apple Metal, or CPU). One install, one import.


## Install

`quantem.widget` is on TestPyPI; it pulls `quantem` and the right GPU backend
automatically (cupy on Linux, Metal on macOS - you never pick one).

```bash
# 1. a fresh environment (conda or mamba)
conda create -n quantem python=3.13 -y
conda activate quantem

# 2. install quantem.widget (auto-pulls quantem + the GPU backend)
pip install -U -i https://test.pypi.org/simple/ --extra-index-url https://pypi.org/simple "quantem.widget>=0.0.1rc0" huggingface_hub
```

Already have a `quantem` environment? Just run step 2 into it.


## Try it on a sample dataset

Downloads a small pre-binned 4D-STEM dataset from Hugging Face (~25 MB), then opens it.


````python
import numpy as np
from quantem.widget import load, Show4DSTEM
from quantem.widget.io import download

path = download('gold_512_npy_bin4')          # ~25 MB from Hugging Face
data = np.load(f'{path}/data.npy')            # (512, 512, 48, 48) 4D-STEM
Show4DSTEM(data)
````

Drag the **detector ROI** (left, diffraction) -> the **virtual image** (right) updates
live. Drag the **scan cursor** on the virtual image -> the diffraction pattern follows.
