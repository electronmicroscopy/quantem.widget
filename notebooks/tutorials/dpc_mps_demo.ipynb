# 4D-STEM on MacBook (MPS) - virtual images + DPC, no binning

One container, one API. Apple Metal on MacBook, torch on CUDA - no backend awareness,
no binning, no quantem dependency. Bit-exact vs quantem.live (parity-tested).


## Environment


```python
from quantem.widget import load, Dataset4dstemGPU, Show2D, device_info
device_info()                       # version, date, GPU / MacBook chip
```

## Load (no bin) -> one GPU dataset


```python
ds = Dataset4dstemGPU(load("path/to/your_master.h5", verbose=True))
ds                                  # Dataset4dstemGPU(512x512x192x192, backend=mps)
```

## Virtual images  (auto probe-fit: BF / ABF / ADF / HAADF / DF / DP)


```python
Show2D(ds.bf())
```

```python
Show2D(ds.adf())
```

## DPC / iDPC  (CoM -> rotation -> integrated phase)


```python
Show2D(ds.idpc())                  # iDPC - projected potential / atoms
```

```python
Show2D(ds.com().col)                  # horizontal CoM - raw DPC field
```
