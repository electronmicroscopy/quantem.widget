# IO.file (multiple files)

Pass a list of file paths to `IO.file([...])` to load them as a stacked 3D dataset. All files must have the same spatial dimensions.

```python
from quantem.widget import IO, Show3D
```

## DM4 files

```python
result = IO.file([
    "/Users/macbook/data/guoliang/widget_demo/gold20nm_RRjacobsens_0020.dm4",
    "/Users/macbook/data/guoliang/widget_demo/gold20nm_RRjacobsens_0021.dm4",
])
print(result)
Show3D(result, title="Gold Nanoparticles")
```
