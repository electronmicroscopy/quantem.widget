# IO.folder (multiple folders)

Pass a list of folder paths to `IO.folder([...])` to combine them into a single stacked dataset. All images must share the same spatial dimensions.

```python
from quantem.widget import IO, Show3D
```

## Two DM3 folders

```python
result = IO.folder(
    [
        "/Users/macbook/data/20251209_KoreanSample/C1_TEM_Bob/Bob_Lee_Sangjoon/",
        "/Users/macbook/data/20251209_KoreanSample/C1_TEM_Bob/Bob_Lee_Sangjoon/20251211_KoreanSample/",
    ],
    file_type="dm3",
)
print(result)
Show3D(result, title="Co3O4 TEM (two sessions)")
```
