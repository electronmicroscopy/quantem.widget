# Field-of-view survey

[![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/bobleesj/quantem.widget/blob/main/docs/tutorials/survey_fov.ipynb)

Use field-of-view grouping when a session contains repeated acquisitions of the same region, such as 0 degree, 90 degree, and EDS files collected from one area.

This view is meant for comparing regions, not just listing files. Images that appear to share the same field of view are grouped into one row. Each HAADF/STEM group is shown as a compact `Show3D` stack so you can flip through the related images, and matching EDS spectrum images appear beside that same row.

Start here when your question is, "Which region should I analyze next?" Use the regular [folder survey](./survey) when you only need a fast inventory of everything in the folder.

````python
from quantem.widget import survey

folder = "/path/to/microscopy-folder"

survey(
    folder,
    thumb=256,
    group_by="fov",
    group_view="stack",
)
````

Review each row as one candidate region. Flip through the image stack, open the matching EDS files, then star or check the files you want to keep for the next analysis step.

Use the selection panel at the bottom to save `.quantem-survey.json`. The file records your choices without deleting or moving raw data, so later notebooks can load only the selected files.

## Command line

Generate a notebook or standalone HTML survey from a terminal. Use the notebook when you want to keep working interactively, and use HTML when you want to share the region survey as a portable review file:

```bash
qw survey /path/to/microscopy-folder --notebook survey_fov.ipynb --thumb 256 --group-by fov --group-view stack
qw survey /path/to/microscopy-folder --html survey_fov.html --thumb 256 --group-by fov --group-view stack
```
