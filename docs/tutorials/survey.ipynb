[![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/bobleesj/quantem.widget/blob/main/docs/tutorials/survey.ipynb)

Use `survey` when you have a folder of microscopy files and want a fast first pass: HAADF/STEM image thumbnails, EDS spectrum-image launchers, an inventory table, and a reversible selection file for follow-up analysis.

The example below expects a local folder of Velox `.emd` files. Replace `folder` with your own data path.

````python
from quantem.widget import survey

folder = "/path/to/microscopy-folder"

survey(folder)
````

Star useful image frames and check useful EDS files, then use the selection panel at the bottom to save `.quantem-survey.json`. The file records choices without deleting or moving raw data. For acquisition sessions with repeated views of the same region, see the FOV survey tutorial.

## Command line

Generate the same notebook or a standalone HTML survey from a terminal:

```bash
qw survey /path/to/microscopy-folder --notebook survey.ipynb --thumb 256
qw survey /path/to/microscopy-folder --html survey.html --thumb 256
```
