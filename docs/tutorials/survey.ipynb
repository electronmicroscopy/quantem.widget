# Folder survey

[![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/bobleesj/quantem.widget/blob/main/docs/tutorials/survey.ipynb)

Use `survey` at the start of a microscopy session review, when you have a folder full of files and want to know what is worth keeping before writing per-file analysis code.

The widget gives you a practical first pass:

- HAADF/STEM images appear first as labeled thumbnails.
- EDS spectrum-image files appear as launchers so you can open each one with `ShowEDS`.
- The inventory table stays nearby for filenames, magnification, scan rotation, shape, and file type.
- Stars and checkboxes let you keep track of useful files without moving or deleting raw data.

The example below expects a local folder of Velox `.emd` files. Replace `folder` with your own data path, run the cell, and start reviewing.

````python
from quantem.widget import survey

folder = "/path/to/microscopy-folder"

survey(folder)
````

After the widget appears, work top to bottom: scan the image gallery, open any EDS files that look relevant, then star useful image frames and check useful EDS files. Use the selection panel at the bottom to save `.quantem-survey.json`; this keeps your review state next to the data without changing the raw files.

When you reopen the same folder, the saved selection is loaded again. That makes the next notebook cleaner: you can continue with only the files you marked during the survey.

If the session contains repeated acquisitions of the same region, such as 0 degree, 90 degree, and EDS files for one field of view, use the [field-of-view survey](./survey_fov) instead.

## Command line

Generate the same review notebook or a standalone HTML survey from a terminal. The HTML file is useful when you want to share the first-pass survey with someone who should not need to run Jupyter:

```bash
qw survey /path/to/microscopy-folder --notebook survey.ipynb --thumb 256
qw survey /path/to/microscopy-folder --html survey.html --thumb 256
```
